```python
import surprise
```

```python
from surprise import SVD
from surprise.model_selection import cross_validate
```

```python
from surprise.dataset import Dataset
```

```python
from surprise.reader import Reader
```

```python
load_from_df=Dataset.load_from_df
```

```python
import pandas as pd
```

```python
df = pd.read_csv('video_ratings.csv')
```

```python
df.head()
df.info()
df.describe()
```

```python
#drop if rating >=5
df2=df[df['rating']<=5]
```

```python
#drop if  rating <0
df2=df2[df2['rating']>0]
```

```python
algo = SVD()
```

```python
df2.info()
```

```python
df3 = pd.DataFrame()
```

```python
df3['user'] = df2['user_id'].values
df3['item'] = df2['movie_id'].values
df3['rating'] = df2['rating'].values
```

```python
data_reader = Reader(line_format='user item rating')
```

```python
data =load_from_df(df3,data_reader)
```

```python
cross_validate(algo, data, measures=['RMSE', 'MAE'], cv=5, verbose=True)
```
