```python
import surprise
```

```python
from surprise import SVD
from surprise.model_selection import cross_validate
```

```python
from surprise.dataset import Dataset
```

```python
from surprise.reader import Reader
```

```python
load_from_df=Dataset.load_from_df
```

```python
import pandas as pd
```

```python
df = pd.read_csv('video_ratings.csv')
```

```python
#hacer un df nuevo con cantidad de veces que un user se repite
df_user = df['user_id'].value_counts().reset_index()
df_user
```

```python
#generar una variable en df_user que sea igual a 1 si count es menor o igual a 3 y 0 si es mayor que 3
df_user['mantener'] = df_user['count'].apply(lambda x: 1 if x <= 3 else 0)
df_user
```

```python
#merge df con df user 
df = pd.merge(df, df_user, on='user_id', how='left')
df
```

```python
#drop si mantener = 0
df = df[df['mantener'] == 1]
df
```

```python
df= df.drop(['mantener'], axis=1)
df
```

```python
df= df.drop(['count'], axis=1)
df
```

```python
algo = SVD()
```

```python
df2 = pd.DataFrame()
```

```python
df2['user'] = df['user_id'].values
df2['item'] = df['movie_id'].values
df2['rating'] = df['rating'].values
```

```python
data_reader = Reader(line_format='user item rating')
```

```python
data =load_from_df(df2,data_reader)
```

```python
cross_validate(algo, data, measures=['RMSE', 'MAE'], cv=5, verbose=True)
```
